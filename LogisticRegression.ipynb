```python
import numpy as np
import pandas as pd
from sklearn import linear_model
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("/content/sample_data/logistic regression dataset-Social_Network_Ads.csv")
df
```

```python
# plt.scatter(df["Age"],df["Purchased"])
plt.scatter(df["EstimatedSalary"],df["Purchased"])
```

```python
x = df[['Age','EstimatedSalary']]
y = df['Purchased']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train)
```

```python
print(y_train)
```

```python
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

```python
# example=np.array([[19,76000]])
# predictv=model.predict(example)
# print(predictv)

example = np.array([[19,76000]])
predictv = model.predict(example)
print(f"Prediction for age 19 and salary 76000: {predictv}")
```

```python
# Example usage with the model's coefficients and intercept
coefficients = model.coef_
intercept = model.intercept_
print(f"Coefficients: {coefficients}")
print(f"Intercept: {intercept}")
```

```python
# Example input
age = 19
salary = 76000

# age = 46
# salary = 41000

def sigmoid(z):
  return 1 / (1 + np.exp(-z))

# Calculate z (the linear combination of features)
z = np.dot(coefficients, [age, salary]) + intercept # This line is added to calculate z

# Apply the sigmoid function
probability = sigmoid(z)

print(f"Probability of purchase for age {age} and salary {salary}: {probability}")
```
